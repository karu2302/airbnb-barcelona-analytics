```python
from pyspark.sql.functions import current_timestamp

spark.sql("CREATE SCHEMA IF NOT EXISTS airbnb_barcelona_project.bronze")

base = "/Volumes/airbnb_barcelona_project/default/raw_data"

listings = (spark.read
    .option("header", True).option("inferSchema", True)
    .option("multiLine", True).option("escape", '"')
    .csv(f"{base}/listings.csv")
    .withColumn("_ingested_at", current_timestamp()))

calendar = (spark.read
    .option("header", True).option("inferSchema", True)
    .csv(f"{base}/calendar.csv")
    .withColumn("_ingested_at", current_timestamp()))

reviews = (spark.read
    .option("header", True).option("inferSchema", True)
    .option("multiLine", True).option("escape", '"')
    .csv(f"{base}/reviews.csv")
    .withColumn("_ingested_at", current_timestamp()))

listings.write.format("delta").mode("overwrite").saveAsTable("airbnb_barcelona_project.bronze.listings")
calendar.write.format("delta").mode("overwrite").saveAsTable("airbnb_barcelona_project.bronze.calendar")
reviews.write.format("delta").mode("overwrite").saveAsTable("airbnb_barcelona_project.bronze.reviews")

print("Bronze tables written.")
print(f"Listings: {spark.table('airbnb_barcelona_project.bronze.listings').count()}")
print(f"Calendar: {spark.table('airbnb_barcelona_project.bronze.calendar').count()}")
print(f"Reviews: {spark.table('airbnb_barcelona_project.bronze.reviews').count()}")
```
