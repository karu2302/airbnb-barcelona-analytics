```python
listings = spark.read.option("header", True).option("inferSchema", True).option("multiLine", True).option("escape", '"').csv(
    "/Volumes/airbnb_barcelona_project/default/raw_data/listings.csv"
)
print(f"Listings: {listings.count()} rows, {len(listings.columns)} columns")
display(listings.limit(5))
listings.printSchema()
```
